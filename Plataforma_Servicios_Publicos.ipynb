````python
from pyspark.sql.functions import col, lit, regexp_replace, when, concat, lpad, create_map
from itertools import chain
from pyspark.sql.types import DoubleType

# 6. silver_tarifa — de EPM tarifas, limpiar $ y coma decimal
df_epm = spark.read.csv(
    "/Volumes/proyecto_integrador/bronze_proyecto_integrador/raw_data/Histórico_de_Tarifas_de_Energía_Eléctrica_–_EPM_(2016-2021)_20261007.csv",
    sep=",", header=True, inferSchema=True
)

# Función para limpiar valores monetarios: quitar $, quitar puntos (separador de miles), reemplazar coma por punto decimal
limpiar_moneda = lambda c: regexp_replace(
    regexp_replace(regexp_replace(col(c), r"\$", ""), r"\.", ""), ",", "."
).cast(DoubleType())

# Mapeo de nombre de mes a numero (pares clave-valor alternados)
meses = {
    "Enero": 1, "Febrero": 2, "Marzo": 3, "Abril": 4, "Mayo": 5, "Junio": 6,
    "Julio": 7, "Agosto": 8, "Septiembre": 9, "Octubre": 10, "Noviembre": 11, "Diciembre": 12
}
mes_map = create_map([lit(x) for x in chain.from_iterable(meses.items())])

silver_tarifa = (
    df_epm
    .filter(col("Año").isNotNull())  # eliminar fila metadato
    .select(
        col("Tipo de Información").alias("tipo_informacion"),
        col("Tipo de Tarifa").alias("tipo_tarifa"),
        col("Tipo de Dato").alias("tipo_dato"),
        col("Nivel").alias("nivel"),
        limpiar_moneda("Propiedad Epm").alias("propiedad_epm"),
        limpiar_moneda("Compartido").alias("compartido"),
        limpiar_moneda("Propiedad del Cliente").alias("propiedad_cliente"),
        limpiar_moneda("Punta").alias("punta"),
        limpiar_moneda("Fuera de Punta").alias("fuera_punta"),
        col("Año").cast("int").alias("anio"),
        mes_map[col("Mes")].alias("mes_num"),
        col("Mes").alias("mes_nombre"),
    )
    .withColumn("periodo", concat(col("anio"), lit("-"), lpad(col("mes_num"), 2, "0")))
)

silver_tarifa.write.mode("overwrite").saveAsTable("proyecto_integrador.silver_proyecto_integrador.silver_tarifa")

print(f"silver_tarifa: {silver_tarifa.count()} filas")
display(spark.table("proyecto_integrador.silver_proyecto_integrador.silver_tarifa").filter(col("periodo") == "2021-01").limit(10))
````

````python
from pyspark.sql.functions import col, hour, dayofweek, month, year, concat, lpad, lit, when

# 7. silver_tiempo — calendario de enero 2021, hora por hora
# 31 dias x 24 horas = 744 filas
silver_tiempo = spark.sql("""
SELECT 
    fecha_hora,
    CAST(fecha_hora AS DATE) AS fecha,
    HOUR(fecha_hora) AS hora,
    DAYOFWEEK(fecha_hora) AS dia_semana,
    MONTH(fecha_hora) AS mes,
    YEAR(fecha_hora) AS anio
FROM (
    SELECT explode(sequence(
        to_timestamp('2021-01-01 00:00:00'),
        to_timestamp('2021-01-31 23:00:00'),
        interval 1 hour
    )) AS fecha_hora
)
""")

silver_tiempo = (silver_tiempo
    .withColumn("periodo", concat(col("anio"), lit("-"), lpad(col("mes"), 2, "0")))
    .withColumn("es_finde", when(col("dia_semana").isin(1, 7), lit(True)).otherwise(lit(False)))
)

silver_tiempo.write.mode("overwrite").saveAsTable("proyecto_integrador.silver_proyecto_integrador.silver_tiempo")

print(f"silver_tiempo: {silver_tiempo.count()} filas")
display(spark.table("proyecto_integrador.silver_proyecto_integrador.silver_tiempo").limit(10))
````

Capa de negocio lista para análisis, siguiendo el modelo estrella. Cada tabla lee de Silver y guarda en `proyecto_integrador.gold_proyecto_integrador`:

- `dim_geografia` — dimensión de municipios de Antioquia (de `silver_geografia`)
- `dim_tiempo` — dimensión temporal de enero 2021, hora a hora (de `silver_tiempo`)
- `hecho_clima` — hecho horario del clima de Medellín, enriquecido con dimensiones
- `hecho_clima_diario` — agregado diario del clima (lluvia, temperatura, etc.)
- `ctx_poblacion` — contexto poblacional por municipio y año
- `ctx_sui_suscriptores` — contexto de suscriptores EPM (Medellín 2021-01)
- `ctx_tarifa` — contexto de tarifas EPM por periodo
- `hecho_resumen_mensual` — tabla de calibración del simulador: Medellín 2021-01 con población, suscriptores, clima y tarifas en una sola fila

````python
%sql
-- 8. dim_geografia — dimensión de municipios de Antioquia
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.dim_geografia AS
SELECT
    cod_divipola,
    nombre_municipio,
    cod_departamento,
    nombre_departamento,
    tipo,
    latitud,
    longitud
FROM proyecto_integrador.silver_proyecto_integrador.silver_geografia;

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.dim_geografia;
````

````python
%sql
-- 9. dim_tiempo — dimensión temporal (enero 2021, hora a hora)
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.dim_tiempo AS
SELECT
    fecha_hora,
    fecha,
    hora,
    dia_semana,
    mes,
    anio,
    periodo,
    es_finde
FROM proyecto_integrador.silver_proyecto_integrador.silver_tiempo;

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.dim_tiempo;
````

````python
%sql
-- 10. hecho_clima — hecho horario, enriquecido con las dimensiones
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.hecho_clima AS
SELECT
    c.cod_divipola,
    g.nombre_municipio,
    t.fecha,
    t.hora,
    t.dia_semana,
    t.es_finde,
    t.periodo,
    c.fecha_hora,
    c.temperature_2m,
    c.precipitation_mm,
    c.relative_humidity_2m,
    c.wind_speed_10m,
    c.surface_pressure
FROM proyecto_integrador.silver_proyecto_integrador.silver_clima c
JOIN proyecto_integrador.gold_proyecto_integrador.dim_geografia g
    ON c.cod_divipola = g.cod_divipola
JOIN proyecto_integrador.gold_proyecto_integrador.dim_tiempo t
    ON c.fecha_hora = t.fecha_hora;

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.hecho_clima;
````

````python
%sql
-- 11. hecho_clima_diario — agregado diario del clima de Medellín
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.hecho_clima_diario AS
SELECT
    cod_divipola,
    CAST(fecha_hora AS DATE) AS fecha,
    date_format(CAST(fecha_hora AS DATE), 'yyyy-MM') AS periodo,
    ROUND(AVG(temperature_2m), 2) AS temp_prom_c,
    ROUND(MAX(temperature_2m), 2) AS temp_max_c,
    ROUND(MIN(temperature_2m), 2) AS temp_min_c,
    ROUND(SUM(precipitation_mm), 2) AS lluvia_total_mm,
    ROUND(AVG(relative_humidity_2m), 1) AS humedad_prom_pct,
    ROUND(AVG(wind_speed_10m), 2) AS viento_prom_ms,
    ROUND(AVG(surface_pressure), 1) AS presion_prom_hpa,
    COUNT(*) AS horas_observadas
FROM proyecto_integrador.silver_proyecto_integrador.silver_clima
GROUP BY cod_divipola, CAST(fecha_hora AS DATE), date_format(CAST(fecha_hora AS DATE), 'yyyy-MM')
ORDER BY fecha;

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.hecho_clima_diario;
````

````python
%sql
-- 12. ctx_poblacion — contexto poblacional por municipio y año
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.ctx_poblacion AS
SELECT
    cod_divipola,
    nombre_municipio,
    anio,
    habitantes
FROM proyecto_integrador.silver_proyecto_integrador.silver_poblacion;

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.ctx_poblacion;
````

````python
%sql
-- 13. ctx_sui_suscriptores — suscriptores EPM (Medellín, enero 2021)
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores AS
SELECT
    cod_divipola,
    nombre_departamento,
    empresa,
    anio,
    periodo_mes,
    periodo,
    estrato_1,
    estrato_2,
    estrato_3,
    estrato_4,
    estrato_5,
    estrato_6,
    total_residencial,
    industrial,
    comercial,
    oficial,
    otros,
    total_no_residencial
FROM proyecto_integrador.silver_proyecto_integrador.silver_sui_suscriptores;

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores;
````

````python
%sql
-- 14. ctx_tarifa — tarifas EPM por periodo (clave: periodo, no municipio)
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.ctx_tarifa AS
SELECT
    tipo_informacion,
    tipo_tarifa,
    tipo_dato,
    nivel,
    propiedad_epm,
    compartido,
    propiedad_cliente,
    punta,
    fuera_punta,
    anio,
    mes_num,
    mes_nombre,
    periodo
FROM proyecto_integrador.silver_proyecto_integrador.silver_tarifa;

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.ctx_tarifa;
````

````python
%sql
-- 15. hecho_resumen_mensual — calibración del simulador: Medellín 2021-01
-- Una fila con población, suscriptores, clima del mes y tarifas promedio
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual AS
WITH hab AS (
    SELECT habitantes
    FROM proyecto_integrador.silver_proyecto_integrador.silver_poblacion
    WHERE cod_divipola = '05001' AND anio = 2021
),
sus AS (
    SELECT *
    FROM proyecto_integrador.silver_proyecto_integrador.silver_sui_suscriptores
    WHERE cod_divipola = '05001' AND periodo = '2021-01'
),
clima AS (
    SELECT
        ROUND(AVG(temperature_2m), 2) AS temp_prom_c,
        ROUND(MAX(temperature_2m), 2) AS temp_max_c,
        ROUND(MIN(temperature_2m), 2) AS temp_min_c,
        ROUND(SUM(precipitation_mm), 2) AS lluvia_total_mm,
        ROUND(AVG(relative_humidity_2m), 1) AS humedad_prom_pct,
        ROUND(AVG(wind_speed_10m), 2) AS viento_prom_ms,
        COUNT(*) AS horas_observadas
    FROM proyecto_integrador.silver_proyecto_integrador.silver_clima
),
tar AS (
    SELECT
        ROUND(AVG(propiedad_epm), 2) AS tarifa_prop_epm_prom,
        ROUND(AVG(compartido), 2) AS tarifa_compartido_prom,
        ROUND(AVG(propiedad_cliente), 2) AS tarifa_prop_cliente_prom,
        ROUND(AVG(punta), 2) AS tarifa_punta_prom,
        ROUND(AVG(fuera_punta), 2) AS tarifa_fuera_punta_prom,
        COUNT(*) AS tarifas_disponibles
    FROM proyecto_integrador.silver_proyecto_integrador.silver_tarifa
    WHERE periodo = '2021-01'
)
SELECT
    g.cod_divipola,
    g.nombre_municipio,
    '2021-01' AS periodo,
    h.habitantes,
    s.estrato_1,
    s.estrato_2,
    s.estrato_3,
    s.estrato_4,
    s.estrato_5,
    s.estrato_6,
    s.total_residencial,
    s.industrial,
    s.comercial,
    s.oficial,
    s.otros,
    s.total_no_residencial,
    c.temp_prom_c,
    c.temp_max_c,
    c.temp_min_c,
    c.lluvia_total_mm,
    c.humedad_prom_pct,
    c.viento_prom_ms,
    c.horas_observadas,
    t.tarifa_prop_epm_prom,
    t.tarifa_compartido_prom,
    t.tarifa_prop_cliente_prom,
    t.tarifa_punta_prom,
    t.tarifa_fuera_punta_prom,
    t.tarifas_disponibles
FROM proyecto_integrador.gold_proyecto_integrador.dim_geografia g
CROSS JOIN hab h
CROSS JOIN sus s
CROSS JOIN clima c
CROSS JOIN tar t
WHERE g.cod_divipola = '05001';

SELECT * FROM proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual;
````

# 🚀 Ampliación a los 125 municipios de Antioquia (2026-10-27)

Objetivo: que la **Vista Ejecutiva por Municipio** del dashboard responda para **cualquier** municipio del filtro global, no solo Medellín.

| Tema | Tratamiento |
| --- | --- |
| Clima | **REAL** — enero 2021 hora por hora para los 125 municipios (Open-Meteo Archive API, coordenadas de cabecera del DIVIPOLA). JSON crudos en `raw_data/open_meteo_antioquia_2021-01/` |
| Suscriptores | **DERIVADO/SIMULADO** — solo existía la fila real SUI de Medellín; los demás se generan de forma DETERMINISTA (HASH del `cod_divipola`) a partir de la población DANE 2021, calibrada con la distribución real de Medellín. La fila de Medellín se recalcula con la misma fórmula (uniformidad, autorizado). Etiquetado en comentario de tabla y columnas |
| `hecho_resumen_mensual` | Una fila por municipio (125): habitantes **reales** DANE, estratos **derivados**, clima **real** por municipio. Sin columnas de tarifa (excluidas del dashboard) |
| Intocables | `ctx_tarifa` y `ctx_poblacion` NO se modifican |

**Determinismo:** todas las fórmulas usan `HASH(cod_divipola)` con constantes fijas → resultados reproducibles ante el jurado.

````python
import urllib.request, json, os, time
from pyspark.sql.functions import col

# 17. Descargar clima REAL de enero 2021 para los 125 municipios de Antioquia
# Open-Meteo Archive API · coordenadas de cabecera municipal desde dim_geografia (DIVIPOLA)
CARPETA = "/Volumes/proyecto_integrador/bronze_proyecto_integrador/raw_data/open_meteo_antioquia_2021-01"
os.makedirs(CARPETA, exist_ok=True)

municipios = (
    spark.table("proyecto_integrador.gold_proyecto_integrador.dim_geografia")
    .where(col("latitud").isNotNull() & col("longitud").isNotNull())
    .select("cod_divipola", "nombre_municipio", "latitud", "longitud")
    .orderBy("cod_divipola")
    .collect()
)
print(f"Municipios a descargar: {len(municipios)}")

ok, fallidos = 0, []
for m in municipios:
    archivo = f"{CARPETA}/{m.cod_divipola}_2021-01.json"
    if os.path.exists(archivo):  # idempotente: no re-descargar JSON ya válidos
        try:
            with open(archivo) as f:
                if len(json.load(f)["hourly"]["time"]) == 744:
                    ok += 1
                    continue
        except Exception:
            pass  # corrupto: re-descargar
    url = (
        "https://archive-api.open-meteo.com/v1/archive"
        f"?latitude={m.latitud}&longitude={m.longitud}"
        "&start_date=2021-01-01&end_date=2021-01-31"
        "&hourly=temperature_2m,precipitation,relative_humidity_2m,wind_speed_10m,surface_pressure"
        "&timezone=America%2FBogota"
    )
    for intento in range(3):
        try:
            with urllib.request.urlopen(url, timeout=60) as r:
                data = json.loads(r.read().decode())
            assert len(data.get("hourly", {}).get("time", [])) == 744, "respuesta incompleta"
            with open(archivo, "w") as f:
                json.dump(data, f)
            ok += 1
            break
        except Exception as e:
            if intento == 2:
                fallidos.append((m.cod_divipola, m.nombre_municipio, str(e)))
            else:
                time.sleep(3)
    time.sleep(0.3)  # cortesía con la API

print(f"JSON válidos: {ok}/{len(municipios)}")
for c, n, e in fallidos:
    print(f"  FALLO {c} ({n}): {e}")
assert not fallidos, f"{len(fallidos)} descargas fallidas"
````

````python
from pyspark.sql.functions import col, regexp_extract, arrays_zip, posexplode, to_timestamp

# 18. silver_clima — explode hourly de los 125 JSON: 125 x 744 = 93.000 filas
CARPETA = "/Volumes/proyecto_integrador/bronze_proyecto_integrador/raw_data/open_meteo_antioquia_2021-01"

raw = spark.read.json(f"{CARPETA}/*.json").withColumn("archivo", col("_metadata.file_path"))

plano = raw.select(
    regexp_extract(col("archivo"), r"(\d{5})_2021-01\.json$", 1).alias("cod_divipola"),
    col("hourly.time").alias("t"),
    col("hourly.temperature_2m").alias("temp"),
    col("hourly.precipitation").alias("prec"),
    col("hourly.relative_humidity_2m").alias("hum"),
    col("hourly.wind_speed_10m").alias("wind"),
    col("hourly.surface_pressure").alias("pres"),
)

silver_clima = plano.select(
    "cod_divipola",
    posexplode(arrays_zip("t", "temp", "prec", "hum", "wind", "pres")).alias("pos", "h"),
).select(
    col("cod_divipola"),
    to_timestamp(col("h.t")).alias("fecha_hora"),
    col("h.temp").cast("double").alias("temperature_2m"),
    col("h.prec").cast("double").alias("precipitation_mm"),
    col("h.hum").cast("int").alias("relative_humidity_2m"),
    col("h.wind").cast("double").alias("wind_speed_10m"),
    col("h.pres").cast("double").alias("surface_pressure"),
)

silver_clima.write.mode("overwrite").saveAsTable(
    "proyecto_integrador.silver_proyecto_integrador.silver_clima"
)
print(f"silver_clima: {silver_clima.count():,} filas")
display(
    spark.table("proyecto_integrador.silver_proyecto_integrador.silver_clima")
    .groupBy("cod_divipola").count().withColumnRenamed("count", "horas")
    .orderBy("cod_divipola").limit(5)
)
````

````python
%sql
-- 19. hecho_clima y hecho_clima_diario — reconstruidos con los 125 municipios
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.hecho_clima AS
SELECT
    c.cod_divipola,
    g.nombre_municipio,
    t.fecha,
    t.hora,
    t.dia_semana,
    t.es_finde,
    t.periodo,
    c.fecha_hora,
    c.temperature_2m,
    c.precipitation_mm,
    c.relative_humidity_2m,
    c.wind_speed_10m,
    c.surface_pressure
FROM proyecto_integrador.silver_proyecto_integrador.silver_clima c
JOIN proyecto_integrador.gold_proyecto_integrador.dim_geografia g
    ON c.cod_divipola = g.cod_divipola
JOIN proyecto_integrador.gold_proyecto_integrador.dim_tiempo t
    ON c.fecha_hora = t.fecha_hora;

CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.hecho_clima_diario AS
SELECT
    cod_divipola,
    CAST(fecha_hora AS DATE) AS fecha,
    date_format(CAST(fecha_hora AS DATE), 'yyyy-MM') AS periodo,
    ROUND(AVG(temperature_2m), 2) AS temp_prom_c,
    ROUND(MAX(temperature_2m), 2) AS temp_max_c,
    ROUND(MIN(temperature_2m), 2) AS temp_min_c,
    ROUND(SUM(precipitation_mm), 2) AS lluvia_total_mm,
    ROUND(AVG(relative_humidity_2m), 1) AS humedad_prom_pct,
    ROUND(AVG(wind_speed_10m), 2) AS viento_prom_ms,
    ROUND(AVG(surface_pressure), 1) AS presion_prom_hpa,
    COUNT(*) AS horas_observadas
FROM proyecto_integrador.silver_proyecto_integrador.silver_clima
GROUP BY cod_divipola, CAST(fecha_hora AS DATE), date_format(CAST(fecha_hora AS DATE), 'yyyy-MM');

SELECT
    (SELECT COUNT(*) FROM proyecto_integrador.gold_proyecto_integrador.hecho_clima) AS filas_hecho_clima,
    (SELECT COUNT(*) FROM proyecto_integrador.gold_proyecto_integrador.hecho_clima_diario) AS filas_hecho_clima_diario;
````

````python
%sql
-- 20. ctx_sui_suscriptores — 125 municipios · DERIVADO determinista de la población DANE 2021
-- Solo existía la fila REAL de Medellín (SUI). Los demás municipios se generan con fórmulas
-- DETERMINISTAS (HASH del cod_divipola) calibradas con la distribución real de Medellín.
-- La fila de Medellín se recalcula con la MISMA fórmula (uniformidad, autorizado por el usuario).
-- ⚠️ DERIVADO/SIMULADO: NO presentar como dato real del SUI.
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores AS
WITH base AS (
    SELECT
        g.cod_divipola,
        g.nombre_departamento,
        p.habitantes,
        ABS(HASH(g.cod_divipola)) % 1000 AS h0,
        ABS(HASH(CONCAT(g.cod_divipola, '|e1'))) % 1000 AS h1,
        ABS(HASH(CONCAT(g.cod_divipola, '|e2'))) % 1000 AS h2,
        ABS(HASH(CONCAT(g.cod_divipola, '|e3'))) % 1000 AS h3,
        ABS(HASH(CONCAT(g.cod_divipola, '|e4'))) % 1000 AS h4,
        ABS(HASH(CONCAT(g.cod_divipola, '|e5'))) % 1000 AS h5,
        ABS(HASH(CONCAT(g.cod_divipola, '|e6'))) % 1000 AS h6,
        ABS(HASH(CONCAT(g.cod_divipola, '|ind'))) % 1000 AS hind,
        ABS(HASH(CONCAT(g.cod_divipola, '|com'))) % 1000 AS hcom,
        ABS(HASH(CONCAT(g.cod_divipola, '|ofi'))) % 1000 AS hofi,
        ABS(HASH(CONCAT(g.cod_divipola, '|otr'))) % 1000 AS hotr
    FROM proyecto_integrador.gold_proyecto_integrador.dim_geografia g
    JOIN proyecto_integrador.gold_proyecto_integrador.ctx_poblacion p
        ON g.cod_divipola = p.cod_divipola AND p.anio = 2021
),
der AS (
    SELECT
        cod_divipola,
        nombre_departamento,
        -- penetración residencial calibrada con Medellín real (889541/2507620 = 0.3548) ±10 %
        habitantes * 0.3548 * (0.90 + 0.20 * h0 / 999.0) AS residencial_raw,
        -- pesos por estrato calibrados con Medellín real ±25 %
        0.13997 * (0.75 + 0.50 * h1 / 999.0) AS w1,
        0.33615 * (0.75 + 0.50 * h2 / 999.0) AS w2,
        0.28605 * (0.75 + 0.50 * h3 / 999.0) AS w3,
        0.11340 * (0.75 + 0.50 * h4 / 999.0) AS w4,
        0.08088 * (0.75 + 0.50 * h5 / 999.0) AS w5,
        0.04363 * (0.75 + 0.50 * h6 / 999.0) AS w6,
        -- no residencial calibrado con Medellín real (7110/75571/1814/516 sobre 889541) ±25 %
        0.00799 * (0.75 + 0.50 * hind / 999.0) AS wi,
        0.08495 * (0.75 + 0.50 * hcom / 999.0) AS wc,
        0.00204 * (0.75 + 0.50 * hofi / 999.0) AS wo,
        0.00058 * (0.75 + 0.50 * hotr / 999.0) AS wx
    FROM base
),
calc AS (
    SELECT
        cod_divipola,
        nombre_departamento,
        ROUND(residencial_raw * w1 / (w1 + w2 + w3 + w4 + w5 + w6)) AS e1,
        ROUND(residencial_raw * w2 / (w1 + w2 + w3 + w4 + w5 + w6)) AS e2,
        ROUND(residencial_raw * w3 / (w1 + w2 + w3 + w4 + w5 + w6)) AS e3,
        ROUND(residencial_raw * w4 / (w1 + w2 + w3 + w4 + w5 + w6)) AS e4,
        ROUND(residencial_raw * w5 / (w1 + w2 + w3 + w4 + w5 + w6)) AS e5,
        ROUND(residencial_raw * w6 / (w1 + w2 + w3 + w4 + w5 + w6)) AS e6,
        ROUND(residencial_raw * wi) AS industrial,
        ROUND(residencial_raw * wc) AS comercial,
        ROUND(residencial_raw * wo) AS oficial,
        ROUND(residencial_raw * wx) AS otros
    FROM der
)
SELECT
    cod_divipola,
    nombre_departamento,
    'DERIVADO (simulado determinista desde población DANE 2021 — NO es dato real del SUI)' AS empresa,
    2021 AS anio,
    1 AS periodo_mes,
    '2021-01' AS periodo,
    CAST(e1 AS INT) AS estrato_1,
    CAST(e2 AS INT) AS estrato_2,
    CAST(e3 AS INT) AS estrato_3,
    CAST(e4 AS INT) AS estrato_4,
    CAST(e5 AS INT) AS estrato_5,
    CAST(e6 AS INT) AS estrato_6,
    CAST(e1 + e2 + e3 + e4 + e5 + e6 AS INT) AS total_residencial,
    CAST(industrial AS INT) AS industrial,
    CAST(comercial AS INT) AS comercial,
    CAST(oficial AS INT) AS oficial,
    CAST(otros AS INT) AS otros,
    CAST(industrial + comercial + oficial + otros AS INT) AS total_no_residencial,
    'DERIVADO_DE_POBLACION' AS origen
FROM calc;

COMMENT ON TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores IS
    'Suscriptores de energía por municipio, enero 2021. DERIVADO/SIMULADO: generado de forma determinista (HASH de cod_divipola) a partir de ctx_poblacion (DANE 2021), calibrado con la fila real SUI de Medellín (recalculada con la misma fórmula para uniformidad, autorizado). NO presentar como dato real del SUI.';

ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN origen COMMENT 'DERIVADO_DE_POBLACION = valor generado deterministamente, no dato real';
ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN total_residencial COMMENT 'DERIVADO — suma de estratos derivados (simulado determinista)';
ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN estrato_1 COMMENT 'DERIVADO — calibrado con Medellín real, perturbado con HASH';
ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN estrato_2 COMMENT 'DERIVADO — calibrado con Medellín real, perturbado con HASH';
ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN estrato_3 COMMENT 'DERIVADO — calibrado con Medellín real, perturbado con HASH';
ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN estrato_4 COMMENT 'DERIVADO — calibrado con Medellín real, perturbado con HASH';
ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN estrato_5 COMMENT 'DERIVADO — calibrado con Medellín real, perturbado con HASH';
ALTER TABLE proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores
    ALTER COLUMN estrato_6 COMMENT 'DERIVADO — calibrado con Medellín real, perturbado con HASH';

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores;
````

````python
%sql
-- 21. hecho_resumen_mensual — UNA FILA POR MUNICIPIO (125)
-- habitantes REALES (DANE 2021) · estratos/suscriptores DERIVADOS · clima REAL por municipio
-- Sin columnas de tarifa (excluidas del dashboard; ctx_tarifa no se toca).
CREATE OR REPLACE TABLE proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual AS
WITH clima_mes AS (
    SELECT
        cod_divipola,
        ROUND(AVG(temp_prom_c), 2) AS temp_prom_c,
        ROUND(MAX(temp_max_c), 2) AS temp_max_c,
        ROUND(MIN(temp_min_c), 2) AS temp_min_c,
        ROUND(SUM(lluvia_total_mm), 2) AS lluvia_total_mm,
        ROUND(AVG(humedad_prom_pct), 1) AS humedad_prom_pct,
        ROUND(AVG(viento_prom_ms), 2) AS viento_prom_ms,
        SUM(horas_observadas) AS horas_observadas
    FROM proyecto_integrador.gold_proyecto_integrador.hecho_clima_diario
    GROUP BY cod_divipola
)
SELECT
    g.cod_divipola,
    g.nombre_municipio,
    '2021-01' AS periodo,
    p.habitantes,
    s.estrato_1,
    s.estrato_2,
    s.estrato_3,
    s.estrato_4,
    s.estrato_5,
    s.estrato_6,
    s.total_residencial,
    s.industrial,
    s.comercial,
    s.oficial,
    s.otros,
    s.total_no_residencial,
    c.temp_prom_c,
    c.temp_max_c,
    c.temp_min_c,
    c.lluvia_total_mm,
    c.humedad_prom_pct,
    c.viento_prom_ms,
    c.horas_observadas
FROM proyecto_integrador.gold_proyecto_integrador.dim_geografia g
JOIN proyecto_integrador.gold_proyecto_integrador.ctx_poblacion p
    ON g.cod_divipola = p.cod_divipola AND p.anio = 2021
JOIN proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores s
    ON g.cod_divipola = s.cod_divipola
JOIN clima_mes c
    ON g.cod_divipola = c.cod_divipola;

COMMENT ON TABLE proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual IS
    'Resumen mensual por municipio, enero 2021 (125 filas). Habitantes REALES (DANE 2021); estratos y suscriptores DERIVADOS de forma determinista (ver ctx_sui_suscriptores); temp_prom_c y lluvia_total_mm REALES de Open-Meteo por municipio.';

SELECT COUNT(*) AS filas FROM proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual;
````

````python
from pyspark.sql.functions import col, min as smin, max as smax, count as scount

# 22. Validación — ampliación a 125 municipios
esperados = {
    "proyecto_integrador.silver_proyecto_integrador.silver_clima": 93000,
    "proyecto_integrador.gold_proyecto_integrador.hecho_clima": 93000,
    "proyecto_integrador.gold_proyecto_integrador.hecho_clima_diario": 3875,
    "proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores": 125,
    "proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual": 125,
}
todo_ok = True
for tabla, esperado in esperados.items():
    n = spark.table(tabla).count()
    ok = n == esperado
    todo_ok &= ok
    print(f"[{'OK' if ok else 'ERROR'}] {tabla}: {n:,} filas (esperadas {esperado:,})")

# cada municipio debe tener exactamente 31 días de clima diario
res = (
    spark.table("proyecto_integrador.gold_proyecto_integrador.hecho_clima_diario")
    .groupBy("cod_divipola").agg(scount("*").alias("dias"))
    .agg(smin("dias").alias("min_dias"), smax("dias").alias("max_dias"), scount("*").alias("municipios"))
    .collect()[0]
)
print(f"hecho_clima_diario por municipio: min {res['min_dias']} / max {res['max_dias']} días · {res['municipios']} municipios")
todo_ok &= (res['min_dias'] == 31 and res['max_dias'] == 31 and res['municipios'] == 125)

# consistencia interna: la suma de estratos debe ser total_residencial
viol = (
    spark.table("proyecto_integrador.gold_proyecto_integrador.ctx_sui_suscriptores")
    .where(
        col("estrato_1") + col("estrato_2") + col("estrato_3")
        + col("estrato_4") + col("estrato_5") + col("estrato_6") != col("total_residencial")
    )
    .count()
)
print(f"Inconsistencias estratos vs total_residencial: {viol}")
todo_ok &= (viol == 0)

print("\nMuestra — 3 municipios más grandes y 2 más pequeños:")
display(
    spark.table("proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual")
    .select("cod_divipola", "nombre_municipio", "habitantes", "total_residencial", "temp_prom_c", "lluvia_total_mm")
    .orderBy(col("habitantes").desc())
    .limit(3)
)
display(
    spark.table("proyecto_integrador.gold_proyecto_integrador.hecho_resumen_mensual")
    .select("cod_divipola", "nombre_municipio", "habitantes", "total_residencial", "temp_prom_c", "lluvia_total_mm")
    .orderBy(col("habitantes").asc())
    .limit(2)
)
assert todo_ok, "Validación fallida — revisar conteos"
````

````python
from pyspark.sql import functions as F
from pyspark.sql.types import DoubleType, StringType

# 1. Definir nombres de tablas y esquemas
# (los esquemas son SOLO el nombre del schema; el catálogo se antepone una sola vez)
catalog_name = "proyecto_integrador"
silver_schema = "silver_proyecto_integrador"

# Asegurar el uso del catálogo
spark.sql(f"USE CATALOG {catalog_name}")

# 2. Leer el DIVIPOLA crudo de la capa Bronze (no hay tabla bronze: el crudo es el CSV del volumen)
# Encabezados reales: "Código Departamento", "Nombre Departamento", "Código Municipio",
# "Nombre Municipio", "Tipo: ...", "longitud", "Latitud" (lat/lon con decimal coma: "-75,581775")
df_divipola_bronze = spark.read.csv(
    "/Volumes/proyecto_integrador/bronze_proyecto_integrador/raw_data/DIVIPOLA-_Códigos_municipios_20261007.csv",
    sep=",",
    header=True,
    inferSchema=True,
)

# 3. Transformación y Limpieza en la capa Silver
# - Estandarizar nombres de columnas a minúsculas y sin espacios
# - Asegurar tipos de datos correctos para latitudes y longitudes
# - códigos normalizados con lpad (5001 -> "05001"), decimales de coma a punto
df_divipola_silver = (
    df_divipola_bronze
    .select(
        F.lpad(F.col("Código Departamento").cast(StringType()), 2, "0").alias("cod_departamento"),
        F.initcap(F.trim(F.col("Nombre Departamento"))).alias("nombre_departamento"),
        F.lpad(F.col("Código Municipio").cast(StringType()), 5, "0").alias("cod_municipio"),
        F.initcap(F.trim(F.col("Nombre Municipio"))).alias("nombre_municipio"),
        F.regexp_replace(F.col("Latitud"), ",", ".").cast(DoubleType()).alias("latitud"),
        F.regexp_replace(F.col("longitud"), ",", ".").cast(DoubleType()).alias("longitud")
    )
    # Eliminar duplicados si los hubiera
    .dropDuplicates(["cod_municipio"])
)

# 4. Guardar el resultado en la capa Silver como tabla Delta
(
    df_divipola_silver.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(f"{catalog_name}.{silver_schema}.dim_divipola")
)

print("¡Capa Silver de Divipola procesada y guardada exitosamente!")
````

````python
from pyspark.sql import functions as F

# 1. Fuentes REALES del proyecto (las tablas del template fact_fallas_red / fact_clima NO existen
catalog_name = "proyecto_integrador"
gold_schema = "gold_proyecto_integrador"

# Dimensión geográfica: dim_divipola (creada en la celda anterior) filtrada a Antioquia,
# renombrando la llave al estándar del proyecto (cod_divipola, 5 dígitos)
df_divipola = (
    spark.table(f"{catalog_name}.silver_proyecto_integrador.dim_divipola")
    .where(F.col("cod_departamento") == "05")
    .withColumnRenamed("cod_municipio", "cod_divipola")
)

# Resumen por municipio ya unificado: habitantes REALES (DANE) · suscriptores DERIVADOS · clima REAL
df_resumen = spark.table(f"{catalog_name}.{gold_schema}.hecho_resumen_mensual")

# 2. Fallas SIMULADAS deterministas — no existe fuente real de fallas en el proyecto
#    (el dashboard también las simula). Calibrado con el dashboard: Medellín = 169 fallas.
#    ⚠️ SIMULADO: NO presentar como dato real de operación.
df_fallas_sim = (
    df_resumen.select("cod_divipola", "total_residencial")
    .withColumn(
        "total_fallas",
        F.round(
            F.col("total_residencial") * F.lit(0.00015948065630541698)
            * (F.lit(0.65) + F.lit(0.70) * (F.abs(F.hash("cod_divipola")) % 1000) / 999.0)
        ).cast("int"),
    )
    .withColumn(
        "promedio_tiempo_respuesta_horas",
        F.round(F.lit(1.5) + F.lit(5.0) * (F.abs(F.hash("cod_divipola", F.lit("resp"))) % 1000) / 999.0, 2),
    )
    .withColumn("origen_fallas", F.lit("SIMULADO_DE_POBLACION"))
    .drop("total_residencial")
)

# 3. Unir la dimensión geográfica con habitantes, suscriptores, clima y fallas por municipio
df_gold_final = (
    df_divipola
    .join(df_resumen.drop("nombre_municipio"), on="cod_divipola", how="left")
    .join(df_fallas_sim, on="cod_divipola", how="left")
    .na.fill({"total_fallas": 0})
    .orderBy("cod_divipola")
)

# 4. Guardar en la capa Gold para consumo (Power BI)
(
    df_gold_final.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(f"{catalog_name}.{gold_schema}.mab_utilidades_resumen")
)

print("¡Capa Gold generada correctamente para su consumo en Power BI!")
print(f"Municipios: {df_gold_final.count()} · Medellín: "
      f"{df_gold_final.where(F.col('cod_divipola') == '05001').select('total_fallas').collect()[0][0]} fallas simuladas")
````
